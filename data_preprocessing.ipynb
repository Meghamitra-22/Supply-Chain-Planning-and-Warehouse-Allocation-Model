```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("India_Cities.csv")
```

```python
df.head()
```

```python
df = df[['city', 'lat', 'lng', 'population']]
df.columns = [
    'city',
    'latitude',
    'longitude',
    'population'
]
```

```python
df['city'] = (
    df['city']
    .str.encode('latin1', errors='ignore')  # Added errors='ignore' here
    .str.decode('utf-8', errors='ignore')
)
```

```python
df.head(10)
```

```python
df = df.drop_duplicates(
    subset='city'
)
```

```python
df = df[
    df['population'] > 500000
]
```

```python
df = df.sort_values(
    by='population',
    ascending=False
)
```

```python
df = df.reset_index(drop=True)
```

```python
print(len(df.city))
```

```python
def assign_region(lat):

    if lat > 26:
        return "North"

    elif lat < 15:
        return "South"

    elif lat > 20:
        return "Central"

    else:
        return "West/East"
```

```python

def assign_store_type(pop):

    if pop > 10000000:
        return "Hypermarket"

    elif pop > 1500000:
        return "Regional Hub"

    else:
        return "Retail Store"
```

```python
df['store_type'] = (
    df['population']
    .apply(assign_store_type)
)
```

```python
np.random.seed(42)

df['monthly_demand'] = (
    (
        df['population'] / 500
    )
    *
    np.random.uniform(
        0.8,
        1.2,
        size=len(df)
    )
).astype(int)
```

```python
df['demand_variability'] = np.random.choice(
    ['Low','Medium','High'],
    size=len(df),
    p=[0.3,0.5,0.2]
)
```

```python
def assign_priority(store):

    if store == "Hypermarket":
        return "High"

    elif store == "Regional Hub":
        return "Medium"

    else:
        return "Low"
```

```python
def assign_sla(priority):

    if priority == "High":
        return 24

    elif priority == "Medium":
        return 48

    else:
        return 72
```

```python


def assign_region(lat):

    if lat > 26:
        return "North"

    elif lat < 15:
        return "South"

    elif lat > 20:
        return "Central"

    else:
        return "West/East"

df['region'] = (
    df['latitude']
    .apply(assign_region)
)
```

```python
df['priority_level'] = (
    df['store_type']
    .apply(assign_priority)
)
```

```python
df['delivery_sla_hours'] = (
    df['priority_level']
    .apply(assign_sla)
)
```

```python
df['transport_cost_factor'] = np.random.uniform(
    2,
    6,
    size=len(df)
)
```

```python
df['store_id'] = [
    f"S{i+1}"
    for i in range(len(df))
]
```

```python
df = df[
    [
        'store_id',
        'city',
        'latitude',
        'longitude',
        'population',
        'store_type',
        'monthly_demand',
        'demand_variability',
        'priority_level',
        'region',
        'delivery_sla_hours',
        'transport_cost_factor'
    ]
]
```

```python

print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nStore Type Distribution:")
print(df['store_type'].value_counts())
df.to_csv(
    "cleaned_supply_chain_data.csv",
    index=False
)

```
